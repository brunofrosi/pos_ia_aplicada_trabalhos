## Questão 2 – Atendimentos Hospitalares

- Pacientes passam inicialmente na triagem e ganham um nível de urgência: baixo, médio e alto;
- Quanto maior o nível de urgência, maior a prioridade do paciente;
- Pacientes idosos (idade >= 60) tem prioridade sobre os demais;
- Pacientes são atendidos em ordem de chegada, respeitando suas respectivas prioridades;

Implemente uma simulação do hospital, com a chegada de 10 pacientes com idade e nível de urgência aleatórios. Ainda, simule que o hospital possui apenas um médico em atendimento. Exiba a ordem de atendimento de cada paciente de acordo com as regras definidas.

**Regras:**
Simulação de atendimento prioritário em um hospital com um único médico.
- Dentro do mesmo nível de urgência, pacientes idosos (idade >= 60) têm prioridade.
- Dentro da mesma prioridade, a ordem de chegada é respeitada (Fila - FIFO).
- São simulados 10 pacientes com idade e urgência aleatórias.

### Estrutura

Os pacientes são representados por objetos da classe `Paciente`.

Para organizar o atendimento, é utilizada uma única fila, representada por uma lista. Todos os pacientes entram na fila pela ordem de chegada.

Cada paciente recebe uma prioridade de acordo com a urgência e a idade:
- Alta + idoso          → prioridade 1
- Alta + não idoso      → prioridade 2
- Média + idoso         → prioridade 3
- Média + não idoso     → prioridade 4
- Baixa + idoso         → prioridade 5
- Baixa + não idoso     → prioridade 6

Na hora do atendimento, o sistema procura na fila o paciente de maior prioridade. Em caso de mesma prioridade, é mantida a ordem de chegada.


```python
# importa a biblioteca para gerar aleatoriamente a idade e a urgencia dos pacientes
import random

# classe utilizada para representar cada paciente
class Paciente:
    # metodo que inicializa um novo paciente
    # __init__ é um metodo especial que é chamado quando a classe é instanciada
    # self é uma referencia ao objeto que esta sendo instanciado
    def __init__(self, nome, idade, urgencia, ordem_chegada):
        self.nome = nome
        self.idade = idade
        self.urgencia = urgencia
        self.ordem_chegada = ordem_chegada

    # metodo que verifica se o paciente idoso
    def idoso(self):
        return self.idade >= 60

    # calcula a prioridade do paciente
    # quanto menor o número, maior a prioridade
    
    # Alta  + idoso     = 1
    # Alta  + não idoso = 2
    # Média + idoso     = 3
    # Média + não idoso = 4
    # Baixa + idoso     = 5
    # Baixa + não idoso = 6
    def prioridade(self):
        if self.urgencia == "Alta":
            prioridade = 1
        elif self.urgencia == "Média":
            prioridade = 3
        else:
            prioridade = 5

        # nao idosos recebem uma prioridade menor
        if not self.idoso():
            prioridade += 1

        return prioridade

    # sobrescreve o metodo padrão, ao usar print()
    def __str__(self):
        faixa = "Idoso" if self.idoso() else "Não idoso"
        return (f"{self.nome} | Idade: {self.idade} | "
                f"Urgência: {self.urgencia} | {faixa} | "
                f"Prioridade: {self.prioridade()} | "
                f"Ordem Chegada: {self.ordem_chegada}")
```

```python
# criando uma lista para representar uma fila
# a lista funciona como uma fila FIFO: o paciente entra no final
fila = []

# simula 10 pacientes
# idade e urgência são geradas aleatoriamente

urgencias = ["Baixa", "Média", "Alta"]

for i in range(1, 11):
    idade = random.randint(18, 90)
    urgencia = random.choice(urgencias)

    # cria o objeto paciente
    paciente = Paciente(f"Paciente {i}", idade, urgencia, i)

    # adiciona o paciente ao final da fila
    fila.append(paciente)

# mostra a fila inicial
print("ORDEM DE CHEGADA:")
for paciente in fila:
    print(paciente)
```

```python
# atendimento com apenas um medico
ordem_atendimento = []

while fila:
    # começa considerando o primeiro paciente da fila
    paciente_prioritario = fila[0]

    # procura o paciente de maior prioridade
    # em caso de empate, a ordem de chegada permanece
    for paciente in fila:
        if paciente.prioridade() < paciente_prioritario.prioridade():
            paciente_prioritario = paciente

    # remove o paciente escolhido da fila
    fila.remove(paciente_prioritario)
    ordem_atendimento.append(paciente_prioritario)

# mostra a ordem final de atendimento
print("\nORDEM DE ATENDIMENTO:")

# enumerate adiciona numeracao
for i, paciente_atendido in enumerate(ordem_atendimento, start=1):
    print(f"{i}º - {paciente_atendido}")
```
